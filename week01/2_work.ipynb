This notebook analyzes the supplied illustrative teaching dataset. Run the cells from top to bottom.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('teaching_prices_clean.csv')
df['date'] = pd.to_datetime(df['date'])
print('Shape:', df.shape)
print('Columns:', df.columns.tolist())
print('Distinct items:', df['item'].nunique())
print('Distinct dates:', df['date'].nunique())
print('Distinct cities:', df['city'].nunique())
display(df.head())
```

## 1. Check the data
Inspect missing values, duplicates, and the teaching-data label before interpreting results.

```python
print('Missing values by column:')
display(df.isna().sum().to_frame('missing_count'))
print('Duplicate rows:', df.duplicated().sum())
print('Data labels:')
display(df['data_label'].value_counts())
print('Date range:', df['date'].min().date(), 'to', df['date'].max().date())
```

## 2. Claim 1 — Compare the same items between cities
Calculate each item's average price in each city and the difference (City 2 minus City 1).

```python
city_means = df.groupby(['item', 'city'])['price'].mean().unstack()
city_means['City 2 minus City 1'] = city_means['City 2'] - city_means['City 1']
display(city_means.round(2))

ax = city_means[['City 1', 'City 2']].plot(kind='bar', figsize=(9, 5))
ax.set_title('Average recorded price by item and city')
ax.set_ylabel('Recorded price (PKR; illustrative)')
ax.set_xlabel('Tracked item')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 3. Claim 2 — Compare price variability by item
Standard deviation measures how spread out the recorded prices are around each item's mean. A larger value indicates greater variation in the recorded price series.

```python
item_stats = df.groupby('item')['price'].agg(['count', 'mean', 'std', 'min', 'max'])
item_stats = item_stats.sort_values('std', ascending=False)
display(item_stats.round(2))

ax = item_stats['std'].sort_values(ascending=False).plot(kind='bar', figsize=(8, 4))
ax.set_title('Standard deviation of recorded prices by item')
ax.set_ylabel('Standard deviation (PKR; illustrative)')
ax.set_xlabel('Tracked item')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 4. Overall time pattern
This chart shows the average of all recorded item-city prices at each date. It is descriptive only and is not an official inflation measure.

```python
date_means = df.groupby('date')['price'].mean().sort_index()
display(date_means.to_frame('average_recorded_price').round(2))
ax = date_means.plot(marker='o', figsize=(9, 4))
ax.set_title('Average recorded price across all item-city observations')
ax.set_ylabel('Average recorded price (illustrative PKR)')
ax.set_xlabel('Date')
plt.tight_layout()
plt.show()
```
